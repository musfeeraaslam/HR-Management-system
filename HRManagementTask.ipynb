```python
class Employee:
  def __init__(self,emp_Id,emp_name,emp_department,designation,emp_salary,performance_rating):
    self.emp_Id=emp_Id
    self.emp_name=emp_name
    self.emp_department= emp_department
    self.designation=designation
    self.emp_salary= emp_salary
    self.performance_rating= performance_rating

  def info(self):
    print("Employee Id is -> ",self.emp_Id)
    print("Employee name is -> ",self.emp_name)
    print("Employee department is -> ",self.emp_department)
    print("Employee designation is -> ",self.designation)
    print("Employee salary is -> ",self.emp_salary)
    print("Employee rating is -> ",self.performance_rating)

  def calculate_salary(self):
    return self.emp_salary * 12

  def update_rating(self,new_rating):
    self.performance_rating= new_rating
```

```python
a = input("Enter Id?")
b = input("Enter name?")
c = input("Enter department?")
d = input("Enter designation?")
e = int(input("Enter salary?"))
f = float(input("Enter rating?"))
X= Employee(a,b,c,d,e,f)
```

```python
X.info()
```

```python
annual_salary = X.calculate_salary()
print("Annual Salary is ->", annual_salary)
```

```python
new_rating = float(input("Enter new performance rating: "))
X.update_rating(new_rating)
print("Updated Rating ->", X.performance_rating)
```
